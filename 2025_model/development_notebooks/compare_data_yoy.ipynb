I was trying to understand why I got different predictions / scores using the same code from last year vs. this year. 

It turns out it's because last year I only had the ordinals through day 114 instead of 133

It's probably worth checking the staleness of these ordinals going forward

```python
import pandas as pd
import numpy as np
rankings_2025 = pd.read_csv('../data/MMasseyOrdinals.csv')
rankings_2024 = pd.read_csv('../../2024_model/data/MMasseyOrdinals.csv')
```

```python
last_day = rankings_2025.groupby('Season').agg({'RankingDayNum':'max'}).rename(
                                    columns = {'RankingDayNum': 'last_day'})


end_rankings = rankings_2025.join(last_day, on = 'Season')
end_rankings = end_rankings[ (end_rankings.RankingDayNum == end_rankings.last_day)
                            & (end_rankings.SystemName =='WLK')]
end_rankings = end_rankings.groupby(['TeamID', 'Season']).agg({'OrdinalRank':np.mean})
end_rankings.reset_index(inplace = True)
end_rankings.columns = ['TeamID', 'Season', 'avg_rank']
end_rankings['final_rank'] = 100-4*np.log(end_rankings['avg_rank']+1)-end_rankings['avg_rank']/22

```

```python
last_day = rankings_2024.groupby('Season').agg({'RankingDayNum':'max'}).rename(
                                    columns = {'RankingDayNum': 'last_day'})


end_rankings_2024 = rankings_2024.join(last_day, on = 'Season')
end_rankings_2024 = end_rankings_2024[ (end_rankings_2024.RankingDayNum == end_rankings_2024.last_day)
                            & (end_rankings_2024.SystemName =='WLK')]


end_rankings_2024 = end_rankings_2024.groupby(['TeamID', 'Season']).agg({'OrdinalRank':np.mean})
end_rankings_2024.reset_index(inplace = True)
```
